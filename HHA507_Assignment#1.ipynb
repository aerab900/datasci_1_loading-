## Dataset 2: AI Impact on Jobs & Salaries (CSV)

This dataset contains self-reported salary survey data for AI, machine learning, and data jobs. Each row is one reported salary and there are approximately 72,000 rows.

**Features (columns):**

- **work_year**: the year the salary was reported
- **experience_level**: short code for experience level (EN = Entry, MI = Mid, SE = Senior, EX = Executive)
- **employment_type**: short code for employment type (FT = Full-time, PT = Part-time, CT = Contract, FL = Freelance)
- **job_title**: the job title as the person reported it
- **salary**: the salary amount in its original currency
- **salary_currency**: the currency the salary was paid in (e.g., USD, EUR)
- **salary_in_usd**: the salary converted to U.S. dollars so all rows can be compared
- **employee_residence**: the country where the employee lives
- **remote_ratio**: how much of the job is remote (0 = on-site, 50 = hybrid, 100 = fully remote)
- **company_location**: the country where the company is located
- **company_size**: size of the company (S = small, M = medium, L = large)
- **experience_level_label**: the experience level code written out in words
- **employment_type_label**: the employment type code written out in words (e.g., Full-time)
- **work_mode**: remote_ratio writtein in word format (Remote, Hybrid, Onsite)
- **salary_outlier_flag**: True if the salary is unusually high or low compared to the rest of the data
- **role_family**: a broader job category grouped from job_title
- **isco_group_hint**: an international occupation category

## Dataset 3: College Major ROI (CSV)

This dataset looks at the return on investment (ROI) of a college degree. It compares what graduates earn over 10 years. Each row is a college graduate.

**Features (columns):**

- **grad_id**: a unique ID number for each graduate
- **major**: the graduate's specific major (19 majors listed)
- **major_category**: the broader group the major belongs to (STEM, health, business, social_science, education, humanities, arts)
- **institution_tier**: the type of school attended (ivy_elite, top_public, mid_tier, regional, for_profit)
- **institution_selectivity_pctile**: how selective the school is in admissions, as a percantage (higher percentage = harder to get in)
- **gpa**: the graduate's grade point average on a 0–4 scale
- **had_internship**: 1 = yes. 0 = no.
- **completed_on_time**: 1 = yes, completed on time. 0 = no, not completed on time
- **region**: the U.S. region (northeast, south, midwest, west)
- **net_cost_usd**: the total cost of the degree after financial aid, in U.S. dollars
- **debt_usd**: the amount of student loan debt taken on, in U.S. dollars (0 if none)
- **hs_baseline_10yr_usd**: how much the person would have earned over 10 years with only a high school diploma
- **added_earnings_10yr_usd**: the extra money earned over 10 years because of the degree (the degree's "earnings premium")
- **earnings_10yr_usd**: total earnings over 10 years after graduating, in U.S. dollars
- **net_roi_usd**: the degree's net return in dollars (added earnings minus net cost)
- **roi_pct**: the net ROI as a percent of the net cost
- **positive_roi**: 1 if the degree paid off (net_roi_usd is greater than 0), 0 if not
- **high_roi**: 1 if the graduate's net ROI is in the top 25% of the dataset, 0 if not

```python
df3 = '/content/Data/college_major_roi.csv'
read_df3 = pd.read_csv(df3)
read_df3.head(11)
```

## Dataset 4: Financial Sample (XLSX)

This dataset contains made-up sales and profit data for a company that sells six products in five countries. It is Microsoft's sample file for learning Power BI. There are 700 rows, and each row is one sale record.

**Features (columns):**

- **Segment**: the type of customer (Government, Midmarket, Channel Partners, Enterprise, Small Business)
- **Country**: where the sale happened (Canada, France, Germany, Mexico, United States of America)
- **Product**: the product sold (Carretera, Montana, Paseo, Velo, VTT, Amarilla)
- **Discount Band**: the level of discount given (None, Low, Medium, High)
- **Units Sold**: how many units were sold
- **Manufacturing Price**: the cost to make one unit, in dollars
- **Sale Price**: the price one unit was sold for, in dollars
- **Gross Sales**: total sales before discounts (Units Sold × Sale Price)
- **Discounts**: the dollar amount taken off for discounts
- **Sales**: total sales after discounts (Gross Sales minus Discounts)
- **COGS**: Cost of Goods Sold, the total cost to make the units that were sold
- **Profit**: money made after costs (Sales minus COGS)
- **Date**: the date of the sale
- **Month Number**: the month as a number (1–12)
- **Month Name**: the month written out (e.g., January)
- **Year**: the year of the sale (2013 or 2014)

```python
df4 = '/content/Data/Financial Sample.xlsx'
read_df4 = pd.read_excel(df4)
read_df4.head(11)
```

## Dataset 5: Retail Inventory (XLSX)

This dataset contains made-up inventory records for a retail store. There are 1,000 rows, and each row is one product.

**Features (columns):**

- **Product ID**: a unique code for each product (e.g., PRD-1001)
- **Product Name**: the name of the product
- **Category**: the type of product (e.g., Home & Kitchen, Beauty, Electronics, Grocery)
- **Supplier Name**: the company that supplies the product
- **Unit Cost**: what the store pays for one unit, in dollars
- **Retail Price**: what the store sells one unit for, in dollars
- **Opening Stock**: how many units were in stock at the start of the period
- **Units Sold**: how many units were sold during the period
- **Closing Stock**: how many units were left at the end of the period (Opening Stock minus Units Sold)
- **Reorder Point**: the stock level at which the store should order more
- **Stock Status**: whether the product is In Stock or Low Stock (Low Stock means Closing Stock is at or below the Reorder Point)
